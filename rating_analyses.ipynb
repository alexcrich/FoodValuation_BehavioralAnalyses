## Heterogeneity across items

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Access the shared folder
shared_folder_path = ""

# Collate ratings for each food item
ratings = {}

# Iterate through each subject folder
for subject_folder in os.listdir(shared_folder_path):
    subject_folder_path = os.path.join(shared_folder_path, subject_folder)

    # Check if the item is a directory (subject folder)
    if os.path.isdir(subject_folder_path):
        # Construct the path to the ratings CSV file (e.g., 654_ratings.csv)
        task_file_name = f"{subject_folder}_ratings.csv"
        task_file_path = os.path.join(subject_folder_path, task_file_name)

        # Check if the file exists
        if os.path.exists(task_file_path):
            # Read the task CSV file into a DataFrame
            df = pd.read_csv(task_file_path)

            # Append ratings to the ratings dictionary
            for index, row in df.iterrows():
                image = row['image']
                response = row['response']
                if pd.notna(image):  # Ensure 'image' is not NaN
                    if image in ratings:
                        ratings[image].append(response)
                    else:
                        ratings[image] = [response]
        else:
            print(f"No data found for participant {subject_folder}.")

# Ensure there is data
if not ratings:
    print("No ratings data found.")
else:
    # Calculate the average rating and SD for each food item
    average_ratings = {image: sum(responses) / len(responses) for image, responses in ratings.items()}
    sd_ratings = {image: np.std(responses, ddof=1) for image, responses in ratings.items()}

    # Sort the average ratings in ascending order
    sorted_ratings = sorted(average_ratings.items(), key=lambda x: x[1])

    # Separate the data for plotting
    food_items = [item[0] for item in sorted_ratings]
    average_ratings_values = [item[1] for item in sorted_ratings]
    sd_values = [sd_ratings[item[0]] for item in sorted_ratings]

    # Create a scatter plot with improved aesthetics
    plt.figure(figsize=(14, 8))
    plt.errorbar(range(len(food_items)), average_ratings_values, yerr=sd_values, fmt='o', color='steelblue', ecolor='lightgrey', elinewidth=3, capsize=0)  # light blue color with error bars
    plt.xticks(range(len(food_items)), food_items, rotation='vertical', fontsize=14)  # Increased fontsize for xticks

    # Improved label spacing to avoid overlap
    plt.subplots_adjust(bottom=0.25)

    # Bold and increase size of axes titles and main title
    plt.xlabel('Food Items', fontsize=16, fontweight='bold')
    plt.ylabel('Average Rating', fontsize=16, fontweight='bold')
    plt.title('Average Rating for Each Food Item with SD', fontsize=18, fontweight='bold')

    # Remove grid lines for a cleaner look
    plt.grid(False)

    # Specify y-axis min and max values to avoid the wonky presets
    y_min = 1 
    y_max = 5  
    y_step = 1  
    plt.ylim([y_min, y_max])
    plt.yticks(range(y_min, y_max + 1, y_step), fontsize=16)  

    # Ensure layout is tight for better spacing
    plt.tight_layout()

    plt.show()
```

```python
import os
import pandas as pd
import ast  # For converting strings to dictionaries

# Access the shared folder
shared_folder_path = ""

# Function to parse the 'response_x' column into separate columns
def parse_response_column(df, column_name='response_x'):
    parsed_df = df.copy()
    
    # Convert the 'response_x' JSON string to separate columns
    def parse_json(row):
        try:
            response_dict = ast.literal_eval(row)
            return pd.Series(response_dict)
        except (ValueError, SyntaxError):
            return pd.Series({"paying": None, "enjoyable": None, "satisfaction": None})
    
    if column_name in parsed_df.columns:
        parsed_columns = parsed_df[column_name].apply(parse_json)
        # Assign parsed columns to the DataFrame
        parsed_df = parsed_df.join(parsed_columns)
    else:
        print(f"Column '{column_name}' does not exist in the DataFrame.")
    
    return parsed_df

# Iterate through each subject folder
for subject_folder in os.listdir(shared_folder_path):
    subject_folder_path = os.path.join(shared_folder_path, subject_folder)
    
    # Check if the item is a directory (subject folder)
    if os.path.isdir(subject_folder_path):
        try:
            # Construct the paths to the ratings files
            pre_file_name = f"{subject_folder}_subset_ratings_pre.csv"
            post_file_name = f"{subject_folder}_subset_ratings_post.csv"
            
            pre_file_path = os.path.join(subject_folder_path, pre_file_name)
            post_file_path = os.path.join(subject_folder_path, post_file_name)

            # Processing pre file if exists
            if os.path.exists(pre_file_path):
                print(f"Processing 'pre' data for participant {subject_folder}")
                # Read the pre CSV file into a DataFrame
                df_pre = pd.read_csv(pre_file_path)
                
                # Check if parsed columns already exist
                if 'paying' not in df_pre.columns or 'enjoyable' not in df_pre.columns or 'satisfaction' not in df_pre.columns:
                    # Parse the 'response_x' column into separate columns
                    df_pre = parse_response_column(df_pre)
                    
                    # Store the modified DataFrame back to the file
                    df_pre.to_csv(pre_file_path, index=False)
                    
                    print(f"Pre data parsed and saved successfully for participant {subject_folder}.")
                else:
                    print(f"Pre data for participant {subject_folder} already has parsed columns.")
            
            # Processing post file if exists
            if os.path.exists(post_file_path):
                print(f"Processing 'post' data for participant {subject_folder}")
                # Read the post CSV file into a DataFrame
                df_post = pd.read_csv(post_file_path)
                
                # Check if parsed columns already exist
                if 'paying' not in df_post.columns or 'enjoyable' not in df_post.columns or 'satisfaction' not in df_post.columns:
                    # Parse the 'response_x' column into separate columns
                    df_post = parse_response_column(df_post)
                    
                    # Store the modified DataFrame back to the file
                    df_post.to_csv(post_file_path, index=False)
                    
                    print(f"Post data parsed and saved successfully for participant {subject_folder}.")
                else:
                    print(f"Post data for participant {subject_folder} already has parsed columns.")

        except Exception as e:
            print(f"An error occurred while processing participant {subject_folder}: {e}")
```

# Pre-post task for task-eligible participants (n=279): changes in subjective valuation measures

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t
import seaborn as sns

# will print significant t test results and plot pre/post ratings for both start groups, for each of the 3 subjective value scales (willingness to pay, enjoyability, satisfaction)

# =========================
# Stats helpers
# =========================

def paired_stats(pre, post, alpha=0.05):
    """
    Paired stats using diff = post - pre.
    Returns t, df, p, Cohen's dz, and a 95% CI for dz derived from CI of mean(diff).
    """
    pre = np.asarray(pre, dtype=float)
    post = np.asarray(post, dtype=float)

    # Keep complete pairs only
    mask = np.isfinite(pre) & np.isfinite(post)
    pre = pre[mask]
    post = post[mask]

    n = len(pre)
    out = {"n": n, "df": None, "t": None, "p": None, "dz": None, "dz_ci": (None, None)}
    if n < 2:
        return out

    diff = post - pre
    df = n - 1

    mean_diff = float(np.mean(diff))
    sd_diff = float(np.std(diff, ddof=1))
    se_diff = sd_diff / np.sqrt(n)

    # Guard sd_diff == 0
    if sd_diff == 0:
        t_stat = np.nan
        p_val = np.nan
        dz = np.nan
        dz_ci = (np.nan, np.nan)
    else:
        t_stat = mean_diff / se_diff
        p_val = 2 * (1 - t.cdf(abs(t_stat), df))
        dz = mean_diff / sd_diff

        tcrit = t.ppf(1 - alpha/2, df)
        md_lo = mean_diff - tcrit * se_diff
        md_hi = mean_diff + tcrit * se_diff
        dz_ci = (md_lo / sd_diff, md_hi / sd_diff)

    out.update({"df": df, "t": t_stat, "p": p_val, "dz": dz, "dz_ci": dz_ci})
    return out


def format_p_with_label(p, label="p"):
    if p is None or (isinstance(p, float) and np.isnan(p)):
        return f"{label} = NA"
    if p < 0.001:
        return f"{label} < .001"
    return f"{label} = {p:.3f}"


def p_adjust(pvals, method="holm"):
    """
    Multiple-comparisons correction.
    method:
      - "holm" = Holm-Bonferroni (FWER control)
      - "bh"   = Benjamini–Hochberg (FDR control)
    Returns adjusted p-values (same order as input).
    """
    p = np.asarray(pvals, dtype=float)
    m = len(p)
    out = np.full(m, np.nan, dtype=float)

    ok = np.isfinite(p)
    p_ok = p[ok]
    if p_ok.size == 0:
        return out

    m_ok = p_ok.size
    order = np.argsort(p_ok)
    p_sorted = p_ok[order]

    meth = method.lower()
    if meth in ["holm", "holm-bonferroni", "holm_bonferroni"]:
        adj_sorted = (m_ok - np.arange(m_ok)) * p_sorted
        adj_sorted = np.maximum.accumulate(adj_sorted)          # monotone
        adj_sorted = np.clip(adj_sorted, 0, 1)
    elif meth in ["bh", "fdr", "benjamini-hochberg", "benjamini_hochberg"]:
        adj_sorted = p_sorted * m_ok / (np.arange(1, m_ok + 1))
        adj_sorted = np.minimum.accumulate(adj_sorted[::-1])[::-1]  # monotone
        adj_sorted = np.clip(adj_sorted, 0, 1)
    else:
        raise ValueError("method must be 'holm' or 'bh'")

    adj_ok = np.empty_like(p_ok)
    adj_ok[order] = adj_sorted
    out[ok] = adj_ok
    return out


# =========================
# Data processing
# =========================

def process_data_subject_level(subject_folder_path, pre_file_name, post_file_name,
                               response_column, measure_column,
                               allowed_subject_ids=None, verbose=True):
    """
    Reads subject folders and returns:
      - pre/post dicts for congruent-start and incongruent-start
      - long DF for plotting (subject x preference x time)
    Only processes subjects in allowed_subject_ids (if provided).
    """

    allowed_set = None
    if allowed_subject_ids is not None:
        allowed_set = set(map(str, allowed_subject_ids))

    pre_congruent = {'Disliked': [], 'Liked': []}
    post_congruent = {'Disliked': [], 'Liked': []}
    pre_incongruent = {'Disliked': [], 'Liked': []}
    post_incongruent = {'Disliked': [], 'Liked': []}
    long_rows = []

    processed = 0
    skipped_not_allowed = 0
    skipped_missing_files = 0
    skipped_bad_start = 0
    skipped_missing_cols = 0

    for subject_folder in os.listdir(subject_folder_path):
        subject_folder = str(subject_folder)
        subject_path = os.path.join(subject_folder_path, subject_folder)
        if not os.path.isdir(subject_path):
            continue

        if allowed_set is not None and subject_folder not in allowed_set:
            skipped_not_allowed += 1
            continue

        pre_file_path = os.path.join(subject_path, f"{subject_folder}_{pre_file_name}")
        post_file_path = os.path.join(subject_path, f"{subject_folder}_{post_file_name}")
        if not (os.path.exists(pre_file_path) and os.path.exists(post_file_path)):
            skipped_missing_files += 1
            continue

        # Determine start type from task file
        starting_block = None
        task_file_path = os.path.join(subject_path, f"{subject_folder}_task.csv")
        if os.path.exists(task_file_path):
            df_task = pd.read_csv(task_file_path)
            if 'Trial type' in df_task.columns and not df_task.empty:
                starting_block = df_task['Trial type'].iloc[0]

        if starting_block not in ['aligned', 'unaligned']:
            skipped_bad_start += 1
            continue

        start_type = 'Congruent' if starting_block == 'aligned' else 'Incongruent'
        pre_data = pre_congruent if start_type == 'Congruent' else pre_incongruent
        post_data = post_congruent if start_type == 'Congruent' else post_incongruent

        df_pre = pd.read_csv(pre_file_path)
        df_post = pd.read_csv(post_file_path)

        if not ({measure_column, response_column} <= set(df_pre.columns) and
                {measure_column, response_column} <= set(df_post.columns)):
            skipped_missing_cols += 1
            continue

        # participant-level mean of the TWO disliked items and TWO liked items
        for pref, resp_vals in [('Disliked', [1.0, 2.0]), ('Liked', [4.0, 5.0])]:
            pre_vals = df_pre[df_pre[response_column].isin(resp_vals)][measure_column].astype(float)
            post_vals = df_post[df_post[response_column].isin(resp_vals)][measure_column].astype(float)

            pre_mean = pre_vals.mean() if len(pre_vals) else np.nan
            post_mean = post_vals.mean() if len(post_vals) else np.nan

            # only keep complete pre/post pairs
            if np.isfinite(pre_mean) and np.isfinite(post_mean):
                pre_data[pref].append(float(pre_mean))
                post_data[pref].append(float(post_mean))

                long_rows.append({"Subject": subject_folder, "Start Type": start_type,
                                  "Preference": pref, "Time": "Pre-task", "Measure": float(pre_mean)})
                long_rows.append({"Subject": subject_folder, "Start Type": start_type,
                                  "Preference": pref, "Time": "Post-task", "Measure": float(post_mean)})

        processed += 1

    if verbose:
        if allowed_set is None:
            print(f"Processed subjects (no allowed list applied): {processed}")
        else:
            present_allowed = [sid for sid in sorted(allowed_set)
                               if os.path.isdir(os.path.join(subject_folder_path, sid))]
            missing_allowed = [sid for sid in sorted(allowed_set)
                               if not os.path.isdir(os.path.join(subject_folder_path, sid))]
            print(f"Allowed IDs provided: {len(allowed_set)}")
            print(f"Allowed IDs present as folders: {len(present_allowed)}")
            print(f"Allowed IDs missing as folders: {len(missing_allowed)}")
            if len(missing_allowed) > 0:
                print("Missing allowed IDs (no folder found):", missing_allowed)
            print(f"Processed allowed subjects: {processed}")
            print(f"Skipped (not in allowed list): {skipped_not_allowed}")
            print(f"Skipped (missing pre/post files): {skipped_missing_files}")
            print(f"Skipped (bad/missing start block): {skipped_bad_start}")
            print(f"Skipped (missing required columns): {skipped_missing_cols}")

    return pre_congruent, post_congruent, pre_incongruent, post_incongruent, pd.DataFrame(long_rows)


# =========================
# Summary stats (+ p-correction)
# =========================

def calculate_summary(pre_data, post_data, p_correction="holm"):
    """
    Computes averages/SEMs and paired stats for each preference key.
    Applies multiple-comparisons correction ACROSS preferences within this panel.
    (Here: across ['Disliked','Liked'] for congruent, and separately for incongruent.)
    """
    avg_pre = {k: (np.mean(v) if len(v) else np.nan) for k, v in pre_data.items()}
    avg_post = {k: (np.mean(v) if len(v) else np.nan) for k, v in post_data.items()}
    sem_pre = {k: (np.std(v, ddof=1) / np.sqrt(len(v)) if len(v) > 1 else np.nan) for k, v in pre_data.items()}
    sem_post = {k: (np.std(v, ddof=1) / np.sqrt(len(v)) if len(v) > 1 else np.nan) for k, v in post_data.items()}

    stats = {k: paired_stats(pre_data[k], post_data[k], alpha=0.05) for k in pre_data.keys()}

    keys = list(pre_data.keys())
    raw_ps = [stats[k].get("p", np.nan) for k in keys]
    adj_ps = p_adjust(raw_ps, method=p_correction)

    for k, padj in zip(keys, adj_ps):
        stats[k]["p_adj"] = float(padj) if np.isfinite(padj) else np.nan
        stats[k]["p_adj_method"] = p_correction

    return avg_pre, avg_post, sem_pre, sem_post, stats


# =========================
# Plotting
# =========================

def plot_measure_results(ax, avg_pre, avg_post, sem_pre, sem_post, stats_dict,
                         response_labels, ylabel, title, long_df_slice,
                         y_tick_top=None, headroom_frac=0.25):

    pre_values = [avg_pre[k] for k in response_labels]
    post_values = [avg_post[k] for k in response_labels]
    pre_err = [sem_pre[k] for k in response_labels]
    post_err = [sem_post[k] for k in response_labels]

    paired_palette = sns.color_palette("Paired")
    colors = {
        'Disliked_pre': paired_palette[6],
        'Disliked_post': paired_palette[7],
        'Liked_pre': paired_palette[8],
        'Liked_post': paired_palette[9],
    }

    x = np.arange(len(response_labels))
    width = 0.35
    pre_x = x - width / 2
    post_x = x + width / 2

    ax.bar(pre_x, pre_values, width, label='Pre',
           color=[colors[f'{lab}_pre'] for lab in response_labels],
           yerr=pre_err, capsize=5, zorder=1)
    ax.bar(post_x, post_values, width, label='Post',
           color=[colors[f'{lab}_post'] for lab in response_labels],
           yerr=post_err, capsize=5, zorder=1)

    # Points/lines
    rng = np.random.default_rng(7)
    all_point_vals = []
    for i, pref in enumerate(response_labels):
        dfp = long_df_slice[long_df_slice["Preference"] == pref].copy()
        if dfp.empty:
            continue
        piv = dfp.pivot_table(index=["Subject"], columns="Time", values="Measure", aggfunc="mean")
        piv = piv.dropna(subset=["Pre-task", "Post-task"], how="any")
        if piv.empty:
            continue

        all_point_vals.extend(piv["Pre-task"].to_numpy().tolist())
        all_point_vals.extend(piv["Post-task"].to_numpy().tolist())

        jit = rng.normal(0, 0.03, size=len(piv))
        px = pre_x[i] + jit
        qx = post_x[i] + jit

        ax.plot(np.vstack([px, qx]),
                np.vstack([piv["Pre-task"].to_numpy(), piv["Post-task"].to_numpy()]),
                color="black", alpha=0.15, linewidth=1, zorder=2)
        ax.scatter(px, piv["Pre-task"], s=28, color="black", alpha=0.55, zorder=3)
        ax.scatter(qx, piv["Post-task"], s=28, color="black", alpha=0.55, zorder=3)

    ax.set_xlabel('Food Item Rating', fontsize=20, fontweight='bold')
    ax.set_ylabel(ylabel, fontsize=20, fontweight='bold')
    ax.set_title(title, fontsize=22, fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(response_labels)
    ax.tick_params(axis='both', which='major', labelsize=19)
    ax.grid(False)

    # Y scaling
    candidates = [v for v in pre_values + post_values + all_point_vals if np.isfinite(v)]
    data_max = float(np.max(candidates)) if candidates else 1.0

    if y_tick_top is not None:
        base_top = float(y_tick_top)
        if abs(base_top - round(base_top)) < 1e-9 and base_top >= 1:
            ax.set_yticks(np.arange(0, int(round(base_top)) + 1, 1))
    else:
        base_top = data_max

    y_upper = max(base_top * (1.0 + headroom_frac), data_max * 1.10)
    ax.set_ylim([0, y_upper])

    # Stars between base_top and y_upper
    star_y = base_top + 0.60 * (y_upper - base_top)

    # Print stats + stars (use corrected p if available)
    for i, pref in enumerate(response_labels):
        st = stats_dict.get(pref, {})
        p = st.get("p", None)
        df = st.get("df", None)
        tval = st.get("t", None)
        n = st.get("n", None)
        dz = st.get("dz", None)
        dz_lo, dz_hi = st.get("dz_ci", (None, None))

        if df is not None and tval is not None and p is not None:
            p_adj = st.get("p_adj", np.nan)
            method = st.get("p_adj_method", "NA")

            print(
                f"{title} ({pref}; n={n}): "
                f"t({df}) = {tval:.2f}, "
                f"{format_p_with_label(p, 'p')}, "
                f"{format_p_with_label(p_adj, f'p_{method}')}, "
                f"Cohen's dz = {dz:.2f}, 95% CI for dz = [{dz_lo:.2f}, {dz_hi:.2f}]"
            )

            p_for_stars = p_adj if np.isfinite(p_adj) else p
            significance = "***" if p_for_stars < 0.001 else "**" if p_for_stars < 0.01 else "*" if p_for_stars < 0.05 else ""
            if significance:
                ax.text(x[i], star_y, significance,
                        ha='center', va='bottom', fontsize=20, color='black',
                        zorder=10, clip_on=False)


# =========================
# Run everything
# =========================

shared_folder_path = "/Users/alexrich/Documents/onlinedata_fullsample"
response_column = 'response_y'

valid_subject_ids #define based on task analyses

measures = [
    {'measure_column': 'paying', 'pre_file_name': 'subset_ratings_pre.csv', 'post_file_name': 'subset_ratings_post.csv',
     'ylabel': 'Average Willingness to Pay ($)', 'title': 'Willingness to Pay'},
    {'measure_column': 'enjoyable', 'pre_file_name': 'subset_ratings_pre.csv', 'post_file_name': 'subset_ratings_post.csv',
     'ylabel': 'Average "Enjoyable" Rating', 'title': 'Enjoyable'},
    {'measure_column': 'satisfaction', 'pre_file_name': 'subset_ratings_pre.csv', 'post_file_name': 'subset_ratings_post.csv',
     'ylabel': 'Average "Satisfaction" Rating', 'title': 'Satisfaction'}
]

# Multiple-comparisons correction choice:
#   "holm" = more conservative (FWER)
#   "bh"   = FDR control
P_CORRECTION_METHOD = "holm"

fig, axes = plt.subplots(3, 2, figsize=(18, 24), constrained_layout=True)
response_labels = ['Disliked', 'Liked']

for i, measure in enumerate(measures):
    pre_c, post_c, pre_ic, post_ic, long_df = process_data_subject_level(
        shared_folder_path,
        measure['pre_file_name'],
        measure['post_file_name'],
        response_column,
        measure['measure_column'],
        allowed_subject_ids=valid_subject_ids,
        verbose=(i == 0)
    )

    avg_pre_c, avg_post_c, sem_pre_c, sem_post_c, stats_c = calculate_summary(
        pre_c, post_c, p_correction=P_CORRECTION_METHOD
    )
    avg_pre_i, avg_post_i, sem_pre_i, sem_post_i, stats_i = calculate_summary(
        pre_ic, post_ic, p_correction=P_CORRECTION_METHOD
    )

    long_c = long_df[long_df["Start Type"] == "Congruent"]
    long_i = long_df[long_df["Start Type"] == "Incongruent"]

    # Ratings: keep top labeled tick at 5; WTP: allow higher
    if measure["measure_column"] == "paying":
        y_tick_top = None
        headroom_frac = 0.18
    else:
        y_tick_top = 5
        headroom_frac = 0.25

    plot_measure_results(
        axes[i, 0],
        avg_pre_c, avg_post_c, sem_pre_c, sem_post_c, stats_c,
        response_labels, measure['ylabel'], f"Congruent-start - {measure['title']}",
        long_c,
        y_tick_top=y_tick_top,
        headroom_frac=headroom_frac
    )

    plot_measure_results(
        axes[i, 1],
        avg_pre_i, avg_post_i, sem_pre_i, sem_post_i, stats_i,
        response_labels, measure['ylabel'], f"Incongruent-start - {measure['title']}",
        long_i,
        y_tick_top=y_tick_top,
        headroom_frac=headroom_frac
    )

paired_palette = sns.color_palette("Paired")
custom_legend = [
    plt.Line2D([0], [0], color=paired_palette[8], lw=4),
    plt.Line2D([0], [0], color=paired_palette[9], lw=4),
    plt.Line2D([0], [0], color=paired_palette[6], lw=4),
    plt.Line2D([0], [0], color=paired_palette[7], lw=4),
]
fig.legend(
    custom_legend,
    ['Pre-task (Liked)', 'Post-task (Liked)', 'Pre-task (Disliked)', 'Post-task (Disliked)'],
    fontsize=22, loc='lower center', ncol=4, bbox_to_anchor=(0.5, -0.03)
)

plt.show()
```
